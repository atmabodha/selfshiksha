## **DIY:**

Convert these numbers to z-score (in a new DataFrame) and redo the box plot

```python
# Extract the stock prices of TCS from df_stocks as a list. Call it tcs_price_list

tcs_price_list = list(df_stocks["tcs"])
```

```python
df_stats = df_stocks.describe()
```

```python
df_stats
```

```python
# Compute the mean and standard deviation of the values in this list

tcs_mean = df_stats.loc["mean"]["tcs"]
tcs_std = df_stats.loc["std"]["tcs"]

print("Mean of TCS : ", tcs_mean)
print("STD of TCS : ",tcs_std)
```

```python
# Compute Mean and STD without using DataFrame through NumPy

tcs_mean = np.mean(np.array(tcs_price_list))
tcs_std = np.std(np.array(tcs_price_list))

print("Mean of TCS : ", tcs_mean)
print("STD of TCS : ",tcs_std)
```

```python
# Create a new empty DataFrame. Call it df_stocks_zscore

df_stocks_zscore = pd.DataFrame()

# Create another list with the z-score values

tcs_price_list = list(df_stocks["tcs"])

tcs_mean = np.mean(np.array(tcs_price_list))
tcs_std = np.std(np.array(tcs_price_list))

tcs_zscore_list = []

for price in tcs_price_list:
  zscore = (price-tcs_mean)/tcs_std
  tcs_zscore_list.append(zscore)

# Append this new z-score value list to the new DataFrame created

df_stocks_zscore["tcs"] = tcs_zscore_list
```

```python
fig = px.box(df_stocks_zscore, x="tcs",points="all")
fig.show()
```

```python
# Write a for-loop to do this for all the stocks

# Create a new empty DataFrame. Call it df_stocks_zscore

df_stocks_zscore = pd.DataFrame()

# This for-loop iterates over every column of the df_stocks dataframe
# This outer loop runs for each column in the df_stocks dataframe, but bypass the Date column
for col in df_stocks:

  if col != "Date":
    # extracts the values for the particular column and saves it as a list called price_list
    price_list = list(df_stocks[col])
    print('Column Name : ',col)

    # Find the mean and standard deviation for the price list using NumPy
    mean = np.mean(np.array(price_list))
    std = np.std(np.array(price_list))

    # Initialise an empty zscore_list in each iteration to store the z-score values for that particular stock
    zscore_list = []

    # Iterate over the original prices of the stock, compute the z-score values and append it to the zscore_list
    # This inner loop runs over each row of the dataframe (i.e. each value in price_list for the selected column)
    for price in price_list:
      zscore = (price-mean)/std
      zscore_list.append(zscore)

    # Append this new z-score value list to the new DataFrame created
    df_stocks_zscore[col] = zscore_list
```

```python
# Plot the Box Plot for the z-score values of all the 5 stocks using for-loop

fig = go.Figure()

for col in df_stocks_zscore:
  if col != "Date":
    fig.add_trace(go.Box(y=df_stocks_zscore[col].values, name=df_stocks_zscore[col].name, boxpoints = 'all'))

fig.show()
```

```python
# Plot the Box Plot for the z-score values of all the 5 stocks WITHOUT using for-loop

df_stocks_zscore_melt = pd.melt(df_stocks_zscore)
fig = px.box(df_stocks_zscore_melt, x="variable",y="value",points="all",color="variable")
fig.show()
```

```python
df_stocks_monthly_std
```

## **DIY:**

At the beginning of the year, you had 400 stocks of Asian Paints, 700 stocks of TCS and 1900 stocks of Cipla. Plot the total market value of your stocks on a daily and monthly basis.

```python
df_stocks
```

```python
market_value_list = []

for i in range(len(df_stocks)):

  price_asianpaints = df_stocks.loc[i]["asianpaints"]

  price_tcs = df_stocks.loc[i]["tcs"]

  price_cipla = df_stocks.loc[i]["cipla"]

  # 400 stocks of Asian Paints, 700 stocks of TCS and 1900 stocks of Cipla
  market_value = 400*price_asianpaints + 700*price_tcs + 1900*price_cipla

  market_value_list.append(market_value)

df_market_value = pd.DataFrame()
df_market_value["Personal"] = market_value_list
df_market_value["Date"] = list(df_stocks["Date"])
```

```python
df_market_value
```

```python
fig = px.line(df_market_value, x="Date", y="Personal", title='Personal Stock Market Value per Day')
# fig = px.scatter(df_stocks, x="Date", y="asianpaints", title='Stock Price per Day')
fig.show()
```

# Heatmap (Optional)

```python
x = df_stocks["Date"]

y = list(df_stocks.columns)
y.remove("Date")

z = []
for col in df_stocks.columns:
  if col != "Date":
    z.append(df_stocks[col])
```

```python
trace = go.Heatmap(
   x = x,
   y = y,
   z = z,
   type = 'heatmap',
   colorscale = 'Viridis'
)
data = [trace]
fig = go.Figure(data = data)
fig.show()
```

# Subplots

```python
fig = make_subplots(rows=1, cols=2,specs=[[{"type": "domain"}, {"type": "domain"}]], subplot_titles=("Start Valuation","End Valuation"))

fig.add_trace(
    go.Pie(labels=stock_names, values=stock_value_start, domain=dict(x=[0, 0.5])),
    row=1, col=1)

fig.add_trace(
    go.Pie(labels=stock_names, values=stock_value_end, domain=dict(x=[0.5, 1.0])),
    row=1, col=2)

fig.update_layout(height=600, width=800, title_text="Stock Valuations")
fig.show()
```
